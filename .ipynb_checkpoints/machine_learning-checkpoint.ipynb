```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from underthesea import word_tokenize
from IPython.display import display, Markdown
from scipy.sparse import hstack

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import GridSearchCV
from sklearn.dummy import DummyClassifier
from sklearn.utils.class_weight import compute_class_weight
from sklearn.multiclass import OneVsRestClassifier
from sklearn.svm import SVC

from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, classification_report
```

```python
# =========================
# LOAD DATA
# =========================
data_path = "../data/processed/"

train_df = pd.read_excel(os.path.join(data_path, "train.xlsx"))
dev_df   = pd.read_excel(os.path.join(data_path, "dev.xlsx"))
test_df  = pd.read_excel(os.path.join(data_path, "test.xlsx"))

X_train_raw = train_df["clean_review"]
y_train = train_df["label"]

X_dev_raw = dev_df["clean_review"]
y_dev = dev_df["label"]

X_test_raw = test_df["clean_review"]
y_test = test_df["label"]

print("Train size:", len(X_train_raw))
print("Dev size:", len(X_dev_raw))
print("Test size:", len(X_test_raw))
```

```python
# =========================
# LOAD STOPWORDS
# =========================
stopwords_path = os.path.join("..", "dicts", "vietnamese-stopwords.txt")

with open(stopwords_path, "r", encoding="utf-8") as f:
    stopwords = set([line.strip() for line in f])

print("Number of stopwords:", len(stopwords))
```

```python
# =========================
# PREPROCESS FUNCTION
# =========================
def preprocess_text(text):
    if not isinstance(text, str):
        return ""
    
    # 1. Tách từ (underthesea)
    text = word_tokenize(text, format="text")
    
    # 2. Xóa stopwords
    tokens = text.split()
    tokens = [word for word in tokens if word not in stopwords]
    
    return " ".join(tokens)
```

```python
# =========================
# APPLY PREPROCESSING
# =========================
X_train = X_train_raw.apply(preprocess_text)
X_dev   = X_dev_raw.apply(preprocess_text)
X_test  = X_test_raw.apply(preprocess_text)

# Xem thử kết quả
print("Before:", X_train_raw.iloc[0])
print("After :", X_train.iloc[0])
```

```python
# =========================
# PHÂN TÍCH PHÂN PHỐI CLASS
# (Kiểm tra imbalance trước khi xây dựng model)
# =========================
class_dist = y_train.value_counts().sort_index()
class_names = {0: "Negative", 1: "Neutral", 2: "Positive"}

print("=== Phân phối class trong Train ===")
for cls, count in class_dist.items():
    pct = count / len(y_train) * 100
    print(f"  Class {cls} ({class_names[cls]}): {count:>6} mẫu ({pct:.1f}%)")

# Tính class weights để xử lý imbalance
weights = compute_class_weight('balanced', classes=np.unique(y_train), y=y_train)
class_weight_dict = dict(enumerate(weights))
print("\nClass weights (balanced):", class_weight_dict)

# Visualize
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, (split_name, y_split) in zip(axes, [("Train", y_train), ("Dev", y_dev), ("Test", y_test)]):
    counts = y_split.value_counts().sort_index()
    ax.bar([class_names[c] for c in counts.index], counts.values,
           color=["#e74c3c", "#f39c12", "#2ecc71"])
    ax.set_title(f"Phân phối class - {split_name}")
    ax.set_ylabel("Số mẫu")
    for i, v in enumerate(counts.values):
        ax.text(i, v + 30, str(v), ha='center', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# =========================
# TF-IDF: WORD + CHAR (xử lý OOV & subword)
# =========================

# --- Word TF-IDF (giữ nguyên như cũ) ---
vectorizer_word = TfidfVectorizer(
    max_features=20000,
    ngram_range=(1, 2),
    norm='l2',
    sublinear_tf=True   # log(1+tf) thay vì tf thô → giảm ảnh hưởng từ lặp nhiều
)

# --- Char TF-IDF (bắt subword, OOV, lỗi chính tả) ---
# analyzer='char_wb': character n-gram tôn trọng biên từ
# Ví dụ: "đẹppppp" → vẫn bắt được "đẹp" qua các char n-gram
vectorizer_char = TfidfVectorizer(
    analyzer='char_wb',
    ngram_range=(2, 4),   # bigram đến 4-gram ký tự
    max_features=30000,
    norm='l2',
    sublinear_tf=True
)

# Fit & transform
X_train_word = vectorizer_word.fit_transform(X_train)
X_dev_word   = vectorizer_word.transform(X_dev)
X_test_word  = vectorizer_word.transform(X_test)

X_train_char = vectorizer_char.fit_transform(X_train)
X_dev_char   = vectorizer_char.transform(X_dev)
X_test_char  = vectorizer_char.transform(X_test)

# Ghép Word + Char
X_train_combined = hstack([X_train_word, X_train_char])
X_dev_combined   = hstack([X_dev_word,   X_dev_char])
X_test_combined  = hstack([X_test_word,  X_test_char])

# Giữ lại X_train_tfidf (word only) cho Naive Bayes vì NB cần non-negative
X_train_tfidf = X_train_word
X_dev_tfidf   = X_dev_word
X_test_tfidf  = X_test_word

print("Word TF-IDF shape :", X_train_word.shape)
print("Char TF-IDF shape :", X_train_char.shape)
print("Combined shape    :", X_train_combined.shape)
```

```python
# =========================
# NAIVE BAYES
# (Dùng Word TF-IDF vì NB yêu cầu giá trị không âm)
# =========================
nb_params = {"alpha": [0.01, 0.1, 0.5, 1.0]}

nb_grid = GridSearchCV(
    MultinomialNB(),
    nb_params,
    cv=5,
    scoring="f1_macro",
    n_jobs=-1
)

nb_grid.fit(X_train_tfidf, y_train)
nb_best = nb_grid.best_estimator_

print("Best NB params:", nb_grid.best_params_)
```

```python
# =========================
# VISUALIZE NB
# =========================
alphas = nb_grid.cv_results_['param_alpha'].data
scores = nb_grid.cv_results_['mean_test_score']

plt.figure()
plt.plot(alphas, scores, marker='o')
plt.xlabel("Alpha")
plt.ylabel("Macro F1 (CV)")
plt.title("Naive Bayes - Hyperparameter Tuning")

for i in range(len(alphas)):
    plt.text(alphas[i], scores[i], f"{scores[i]:.3f}")

plt.show()
```

```python

# =========================
# SVM - WORD TF-IDF
# =========================
svm_params = {"C": [0.1, 1, 10]}

svm_grid = GridSearchCV(
    LinearSVC(),
    svm_params,
    cv=5,
    scoring="f1_macro",
    n_jobs=-1
)

svm_grid.fit(X_train_tfidf, y_train)
svm_best = svm_grid.best_estimator_

print("Best SVM (Word) params:", svm_grid.best_params_)
```

```python
# =========================
# VISUALIZE SVM
# =========================
C_values = svm_grid.cv_results_['param_C'].data
scores = svm_grid.cv_results_['mean_test_score']

plt.figure()
plt.plot(C_values, scores, marker='o')
plt.xlabel("C")
plt.ylabel("Macro F1 (CV)")
plt.title("SVM - Hyperparameter Tuning")

for i in range(len(C_values)):
    plt.text(C_values[i], scores[i], f"{scores[i]:.3f}")

plt.show()
```

```python
# =========================
# SVM - WORD + CHAR TF-IDF (xử lý OOV)
# =========================
svm_combined_grid = GridSearchCV(
    LinearSVC(),
    {"C": [0.1, 1, 10]},
    cv=5,
    scoring="f1_macro",
    n_jobs=-1
)

svm_combined_grid.fit(X_train_combined, y_train)
svm_combined_best = svm_combined_grid.best_estimator_

print("Best SVM (Word+Char) params:", svm_combined_grid.best_params_)
```

```python
# =========================
# SVM - CLASS WEIGHTING (chiến lược Neutral)
# Class weighting giúp model chú ý hơn đến class ít mẫu (Neutral)
# =========================
svm_weighted_grid = GridSearchCV(
    LinearSVC(class_weight='balanced'),
    {"C": [0.1, 1, 10]},
    cv=5,
    scoring="f1_macro",
    n_jobs=-1
)

svm_weighted_grid.fit(X_train_combined, y_train)
svm_weighted_best = svm_weighted_grid.best_estimator_

print("Best SVM (Weighted, Word+Char) params:", svm_weighted_grid.best_params_)
```

```python
# =========================
# RANDOM FOREST
# =========================
rf_params = {
    "n_estimators": [20, 50, 100, 200],
    "max_depth": [None, 10, 20]
}

rf_grid = GridSearchCV(
    RandomForestClassifier(random_state=42),
    rf_params,
    cv=5,
    scoring="f1_macro",
    n_jobs=-1
)

rf_grid.fit(X_train_tfidf, y_train)
rf_best = rf_grid.best_estimator_

print("Best RF params:", rf_grid.best_params_)
```

```python
# =========================
# VISUALIZE RANDOM FOREST 
# =========================
results_rf = pd.DataFrame(rf_grid.cv_results_)

# lọc max_depth = None (thực chất là NaN)
filtered = results_rf[results_rf['param_max_depth'].isna()]

# convert kiểu dữ liệu
n_estimators = filtered['param_n_estimators'].astype(int)
scores = filtered['mean_test_score']

plt.figure()
plt.plot(n_estimators, scores, marker='o')
plt.xlabel("n_estimators")
plt.ylabel("Macro F1 (CV)")
plt.title("Random Forest (max_depth=None)")

for i in range(len(n_estimators)):
    plt.text(n_estimators.iloc[i], scores.iloc[i], f"{scores.iloc[i]:.3f}")

plt.show()
```

```python
# =========================
# BASELINE MODELS
# So sánh với baseline đơn giản để xác nhận model thực sự học được
# =========================

# Baseline 1: Luôn dự đoán class nhiều nhất (majority class)
dummy_majority = DummyClassifier(strategy='most_frequent', random_state=42)
dummy_majority.fit(X_train_tfidf, y_train)

# Baseline 2: Dự đoán ngẫu nhiên theo đúng tỷ lệ phân phối của train
dummy_stratified = DummyClassifier(strategy='stratified', random_state=42)
dummy_stratified.fit(X_train_tfidf, y_train)

majority_class = y_train.value_counts().idxmax()
print(f"Majority class: {majority_class} ({class_names[majority_class]})")
print("Baseline Majority   - Accuracy:", round(accuracy_score(y_test, dummy_majority.predict(X_test_tfidf)), 4))
print("Baseline Stratified - Accuracy:", round(accuracy_score(y_test, dummy_stratified.predict(X_test_tfidf)), 4))
```

```python
# =========================
# PROFESSIONAL EVALUATION FUNCTION
# =========================
def evaluate(model, X, y, name="Model", use_combined=False):
    y_pred = model.predict(X)

    acc = accuracy_score(y, y_pred)
    f1  = f1_score(y, y_pred, average="macro")

    # Title đẹp
    display(Markdown(f"## {name} Evaluation"))

    # ===== Summary Table =====
    summary_df = pd.DataFrame({
        "Metric": ["Accuracy", "Macro F1"],
        "Score": [round(acc, 4), round(f1, 4)]
    })

    display(summary_df.style.set_caption("Overall Performance"))

    # ===== Classification Report =====
    report = classification_report(y, y_pred, output_dict=True)
    report_df = pd.DataFrame(report).transpose()

    display(report_df.style
            .set_caption("Classification Report")
            .format("{:.4f}"))

    # ===== Confusion Matrix =====
    cm = confusion_matrix(y, y_pred)
    labels = sorted(list(set(y)))
    label_names = [class_names.get(l, str(l)) for l in labels]

    plt.figure()
    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        xticklabels=label_names,
        yticklabels=label_names
    )

    plt.title(f"Confusion Matrix - {name}")
    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")

    plt.show()

    return {"Model": name, "Accuracy": round(acc, 4), "Macro F1": round(f1, 4)}
```

```python
# =========================
# EVALUATE ON DEV
# =========================
evaluate(nb_best,            X_dev_tfidf,    y_dev, "Naive Bayes (Dev)")
evaluate(svm_best,           X_dev_tfidf,    y_dev, "SVM Word (Dev)")
evaluate(svm_combined_best,  X_dev_combined, y_dev, "SVM Word+Char (Dev)")
evaluate(svm_weighted_best,  X_dev_combined, y_dev, "SVM Word+Char Weighted (Dev)")
evaluate(rf_best,            X_dev_tfidf,    y_dev, "Random Forest (Dev)")
```

```python
# =========================
# EVALUATE ON TEST
# =========================
evaluate(nb_best,            X_test_tfidf,    y_test, "Naive Bayes (Test)")
evaluate(svm_best,           X_test_tfidf,    y_test, "SVM Word (Test)")
evaluate(svm_combined_best,  X_test_combined, y_test, "SVM Word+Char (Test)")
evaluate(svm_weighted_best,  X_test_combined, y_test, "SVM Word+Char Weighted (Test)")
evaluate(rf_best,            X_test_tfidf,    y_test, "Random Forest (Test)")
```

```python
# =========================
# COMPARISON TABLE (bao gồm cả baseline)
# =========================
all_results = []

# --- Baseline ---
for name, model, X in [
    ("Baseline - Majority Class",    dummy_majority,   X_test_tfidf),
    ("Baseline - Stratified Random", dummy_stratified, X_test_tfidf),
]:
    y_pred = model.predict(X)
    all_results.append({
        "Model": name,
        "Accuracy": round(accuracy_score(y_test, y_pred), 4),
        "Macro F1": round(f1_score(y_test, y_pred, average="macro"), 4),
        "Nhóm": "Baseline"
    })

# --- ML Models ---
for name, model, X in [
    ("Naive Bayes",              nb_best,           X_test_tfidf),
    ("SVM (Word TF-IDF)",        svm_best,          X_test_tfidf),
    ("SVM (Word+Char TF-IDF)",   svm_combined_best, X_test_combined),
    ("SVM (Word+Char Weighted)", svm_weighted_best, X_test_combined),
    ("Random Forest",            rf_best,           X_test_tfidf),
]:
    y_pred = model.predict(X)
    all_results.append({
        "Model": name,
        "Accuracy": round(accuracy_score(y_test, y_pred), 4),
        "Macro F1": round(f1_score(y_test, y_pred, average="macro"), 4),
        "Nhóm": "ML Model"
    })

results_df = pd.DataFrame(all_results).sort_values(by="Macro F1", ascending=False).reset_index(drop=True)
display(results_df)
```

```python
# =========================
# VISUALIZATION SO SÁNH TẤT CẢ MODELS
# =========================
fig, ax = plt.subplots(figsize=(12, 5))

models_list  = results_df["Model"]
accuracy_list = results_df["Accuracy"]
f1_list       = results_df["Macro F1"]
groups        = results_df["Nhóm"]

x = np.arange(len(models_list))
width = 0.35

# Màu khác nhau cho baseline vs ML
colors_acc = ["#aaaaaa" if g == "Baseline" else "#3498db" for g in groups]
colors_f1  = ["#cccccc" if g == "Baseline" else "#e74c3c" for g in groups]

bars1 = ax.bar(x - width/2, accuracy_list, width, label="Accuracy", color=colors_acc)
bars2 = ax.bar(x + width/2, f1_list,       width, label="Macro F1", color=colors_f1)

ax.set_xticks(x)
ax.set_xticklabels(models_list, rotation=25, ha='right', fontsize=9)
ax.set_ylabel("Score")
ax.set_title("So sánh tất cả models (Test set)")
ax.set_ylim(0, 0.85)
ax.legend()
ax.axhline(y=0.5, color='gray', linestyle='--', alpha=0.5, label='Ngưỡng 0.5')

# Hiển thị số
for bar in bars1:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.005,
            f"{bar.get_height():.3f}", ha='center', va='bottom', fontsize=7)
for bar in bars2:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.005,
            f"{bar.get_height():.3f}", ha='center', va='bottom', fontsize=7)

# Chú thích vùng baseline
n_baseline = (groups == "Baseline").sum()
ax.axvspan(-0.5, n_baseline - 0.5, alpha=0.05, color='gray', label='Baseline zone')

plt.tight_layout()
plt.show()
```

```python
# =========================
# PHÂN TÍCH ĐẶC BIỆT CLASS NEUTRAL
# So sánh F1 của Neutral trên các model
# =========================
neutral_f1_results = []

for name, model, X in [
    ("Naive Bayes",              nb_best,           X_test_tfidf),
    ("SVM (Word)",               svm_best,          X_test_tfidf),
    ("SVM (Word+Char)",          svm_combined_best, X_test_combined),
    ("SVM (Word+Char Weighted)", svm_weighted_best, X_test_combined),
    ("Random Forest",            rf_best,           X_test_tfidf),
]:
    y_pred = model.predict(X)
    report = classification_report(y_test, y_pred, output_dict=True)
    neutral_f1_results.append({
        "Model": name,
        "Negative F1": round(report.get("0", report.get(0, {})).get("f1-score", 0), 4),
        "Neutral F1":  round(report.get("1", report.get(1, {})).get("f1-score", 0), 4),
        "Positive F1": round(report.get("2", report.get(2, {})).get("f1-score", 0), 4),
    })

neutral_df = pd.DataFrame(neutral_f1_results)
display(Markdown("### F1-score theo từng class (Test set)"))
display(neutral_df)

# Plot
x = np.arange(len(neutral_df))
width = 0.25

plt.figure(figsize=(11, 5))
plt.bar(x - width, neutral_df["Negative F1"], width, label="Negative", color="#e74c3c")
plt.bar(x,         neutral_df["Neutral F1"],  width, label="Neutral",  color="#f39c12")
plt.bar(x + width, neutral_df["Positive F1"], width, label="Positive", color="#2ecc71")

plt.xticks(x, neutral_df["Model"], rotation=20, ha='right', fontsize=9)
plt.ylabel("F1-score")
plt.title("F1-score theo class - Neutral khó nhất")
plt.legend()
plt.ylim(0, 0.95)
plt.tight_layout()
plt.show()
```

```python
# =========================
# SAVE RESULTS
# =========================
results_df.to_excel("model_comparison.xlsx", index=False)
neutral_df.to_excel("class_f1_comparison.xlsx", index=False)
print("Đã lưu kết quả ra file Excel.")
```
