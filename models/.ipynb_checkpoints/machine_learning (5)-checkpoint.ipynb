```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from underthesea import word_tokenize
from IPython.display import display, Markdown
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import GridSearchCV
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score, f1_score, confusion_matrix, classification_report
from sklearn.metrics import ConfusionMatrixDisplay
from scipy.sparse import hstack
```

```python
# =========================
# LOAD DATA
# =========================
data_path = "../data/processed/"

train_df = pd.read_excel(os.path.join(data_path, "train.xlsx"))
dev_df   = pd.read_excel(os.path.join(data_path, "dev.xlsx"))
test_df  = pd.read_excel(os.path.join(data_path, "test.xlsx"))

X_train_raw = train_df["clean_review"]
y_train = train_df["label"]

X_dev_raw = dev_df["clean_review"]
y_dev = dev_df["label"]

X_test_raw = test_df["clean_review"]
y_test = test_df["label"]

print("Train size:", len(X_train_raw))
print("Dev size:", len(X_dev_raw))
print("Test size:", len(X_test_raw))
```

```python
# =========================
# TẠO STOPWORD TỪ CORPUS
# =========================
from sklearn.feature_extraction.text import CountVectorizer

def get_top_n_stopwords(corpus, n):
    """Lấy n từ xuất hiện nhiều nhất trong corpus làm stopwords."""
    cv = CountVectorizer()
    cv.fit(corpus)
    word_counts = pd.DataFrame(
        cv.transform(corpus).sum(axis=0).A1,
        index=cv.get_feature_names_out(),
        columns=['count']
    ).sort_values('count', ascending=False)
    return list(word_counts.index[:n])

# Chỉ fit trên train set — tránh data leakage
print('Đang đếm từ phổ biến nhất trong train set...')
top_200_words = get_top_n_stopwords(X_train_raw, 200)
print(f'Top 10 từ phổ biến nhất: {top_200_words[:10]}')
```

```python
# =========================
# PREPROCESS FUNCTION
# =========================
def preprocess_text(text):
    if not isinstance(text, str):
        return ""
    # Tách từ (underthesea)
    text = word_tokenize(text, format="text")
    return text
```

```python
# =========================
# APPLY PREPROCESSING
# =========================
X_train = X_train_raw.apply(preprocess_text)
X_dev   = X_dev_raw.apply(preprocess_text)
X_test  = X_test_raw.apply(preprocess_text)

# Xem thử kết quả
print("Before:", X_train_raw.iloc[0])
print("After :", X_train.iloc[0])
```

```python
# =========================
# THỬ NGHIỆM CÁC MỨC STOPWORD TỪ CORPUS
# =========================
import warnings
warnings.filterwarnings('ignore')

sw_levels = [0, 50, 70, 100, 120, 150, 170, 200]
results   = []

print('Đang thử nghiệm các mức stopword...')
print('-' * 75)

for n in sw_levels:
    sw    = top_200_words[:n] if n > 0 else None
    label = f'Top {n:3d}' if n > 0 else 'Không xóa'

    # TF-IDF
    vec = TfidfVectorizer(
        ngram_range=(1, 2),
        sublinear_tf=True,
        min_df=2,
        stop_words=sw
    )
    X_tr = vec.fit_transform(X_train)
    X_dv = vec.transform(X_dev)
    X_te = vec.transform(X_test)
    weights_ = compute_sample_weight(class_weight='balanced', y=y_train)

    # ── Naive Bayes: tune alpha trên dev
    best_nb_score, best_nb_alpha, best_nb_model = 0, 0.1, None
    for alpha in [0.01, 0.1, 0.5, 1.0]:
        m = MultinomialNB(alpha=alpha)
        m.fit(X_tr, y_train, sample_weight=weights_)
        s = f1_score(y_dev, m.predict(X_dv), average='macro')
        if s > best_nb_score:
            best_nb_score, best_nb_alpha, best_nb_model = s, alpha, m
    nb_f1  = f1_score(y_test, best_nb_model.predict(X_te), average='macro')
    nb_acc = accuracy_score(y_test, best_nb_model.predict(X_te))

    # ── SVM: tune C trên dev
    best_svm_score, best_C, best_svm_model = 0, 1, None
    for C in [0.1, 1, 10]:
        m = LinearSVC(C=C, class_weight='balanced', random_state=42, max_iter=2000)
        m.fit(X_tr, y_train)
        s = f1_score(y_dev, m.predict(X_dv), average='macro')
        if s > best_svm_score:
            best_svm_score, best_C, best_svm_model = s, C, m
    svm_f1  = f1_score(y_test, best_svm_model.predict(X_te), average='macro')
    svm_acc = accuracy_score(y_test, best_svm_model.predict(X_te))

    results.append({
        'Stopwords'  : label,
        'n_features' : X_tr.shape[1],
        'NB alpha'   : best_nb_alpha,
        'NB F1'      : round(nb_f1, 4),
        'NB Acc'     : round(nb_acc, 4),
        'SVM C'      : best_C,
        'SVM F1'     : round(svm_f1, 4),
        'SVM Acc'    : round(svm_acc, 4),
    })
    print(f'[{label:10s}] features={X_tr.shape[1]:>7,} | '
          f'NB F1={nb_f1:.4f} (alpha={best_nb_alpha}) | '
          f'SVM F1={svm_f1:.4f} (C={best_C})')

# ── Bảng kết quả
df_results = pd.DataFrame(results)
print('\n' + '=' * 75)
print(df_results[['Stopwords','n_features','NB alpha','NB F1','NB Acc','SVM C','SVM F1','SVM Acc']].to_string(index=False))
print('=' * 75)
best_nb_row  = df_results.loc[df_results['NB F1'].idxmax()]
best_svm_row = df_results.loc[df_results['SVM F1'].idxmax()]
print(f"\nNB  tốt nhất : {best_nb_row['Stopwords'].strip():10s} "
      f"(F1={best_nb_row['NB F1']:.4f}, alpha={best_nb_row['NB alpha']})")
print(f"SVM tốt nhất : {best_svm_row['Stopwords'].strip():10s} "
      f"(F1={best_svm_row['SVM F1']:.4f}, C={best_svm_row['SVM C']})")

# Lưu vectorizer và model tốt nhất của SVM để dùng tiếp
best_n_svm = sw_levels[df_results['SVM F1'].idxmax()]
sw_best    = top_200_words[:best_n_svm] if best_n_svm > 0 else None
vectorizer = TfidfVectorizer(ngram_range=(1,2), sublinear_tf=True, min_df=2, stop_words=sw_best)
X_train_tfidf = vectorizer.fit_transform(X_train)
X_dev_tfidf   = vectorizer.transform(X_dev)
X_test_tfidf  = vectorizer.transform(X_test)
print(f'\nVectorizer đã fit lại với mức SW tốt nhất cho SVM: {best_n_svm}')
```

```python
weights = compute_sample_weight(class_weight='balanced', y=y_train)
```

```python
# =========================
# NAIVE BAYES
# =========================
nb_params = {"alpha": [0.01, 0.1, 0.5, 1.0]}
weights = compute_sample_weight(class_weight='balanced', y=y_train)
best_score = 0
best_alpha = None
best_model = None
scores = []
alphas_used = []

weights = compute_sample_weight(class_weight='balanced', y=y_train)
for alpha in nb_params["alpha"]:
    # Khởi tạo mô hình với từng giá trị alpha
    model = MultinomialNB(alpha=alpha)
    
    # Huấn luyện trên Train có trọng số
    model.fit(X_train_tfidf, y_train, sample_weight=weights)
    
    # Dự đoán trên Dev
    y_dev_pred = model.predict(X_dev_tfidf)
    
    # Tính Macro F1 trên Dev
    score = f1_score(y_dev, y_dev_pred, average='macro')
    
    scores.append(score)
    alphas_used.append(alpha)
    
    print(f"alpha = {alpha}, Macro F1 on Dev = {score:.4f}")
    
    # Lưu lại mô hình tốt nhất
    if score > best_score:
        best_score = score
        best_alpha = alpha
        best_model = model

print(f"\nBest alpha = {best_alpha}, Macro F1 on Dev = {best_score:.4f}")
```

```python
# =========================
# VISUALIZE NB
# =========================

plt.figure(figsize=(8, 5))
plt.plot(alphas_used, scores, marker='o', linestyle='-', linewidth=2, markersize=8)

plt.xlabel("Alpha", fontsize=12)
plt.ylabel("Macro F1 (Dev)", fontsize=12)
plt.title("Naive Bayes - Hyperparameter Tuning", fontsize=14)

# ===== tìm điểm tốt nhất =====
best_idx = np.argmax(scores)

# ===== offset tự động (QUAN TRỌNG) =====
y_offset = (max(scores) - min(scores)) * 0.03

# ===== vẽ label =====
for i in range(len(alphas_used)):
    if i == best_idx:
        # BEST SCORE -> dưới điểm
        plt.text(
            alphas_used[i],
            scores[i] - y_offset,
            f"{scores[i]:.3f}",
            ha='center',
            va='top',
            fontsize=10,
            
        )
    else:
        # các điểm khác -> trên điểm
        plt.text(
            alphas_used[i],
            scores[i] + y_offset,
            f"{scores[i]:.3f}",
            ha='center',
            va='bottom',
            fontsize=10
        )

plt.grid(True, linestyle='--', alpha=0.7)

# làm trục Y đẹp hơn
plt.ylim(min(scores) - y_offset*3, max(scores) + y_offset*3)

plt.tight_layout()
plt.show()
```

```python
final_nb = MultinomialNB(alpha=best_alpha)
final_nb.fit(X_train_tfidf, y_train, sample_weight=weights)
```

```python
# =========================
# ĐÁNH GIÁ TRÊN TEST
# =========================

y_test_pred = final_nb.predict(X_test_tfidf)

test_accuracy = accuracy_score(y_test, y_test_pred)
test_f1_macro = f1_score(y_test, y_test_pred, average='macro')

print("\n=== KẾT QUẢ TRÊN TEST ===")
print(f"Accuracy = {test_accuracy:.4f}")
print(f"Macro F1 = {test_f1_macro:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_test_pred, digits=4))
```

```python
# Tính confusion matrix
cm = confusion_matrix(y_test, y_test_pred)

labels = ['Tiêu cực', 'Trung tính', 'Tích cực']


# Vẽ heatmap
plt.figure(figsize=(7, 6))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=labels,
    yticklabels=labels,
    cbar_kws={'label': 'Số lượng mẫu'}
)

plt.xlabel("Predicted Label", fontsize=12)
plt.ylabel("True Label", fontsize=12)
plt.title("Confusion Matrix - Naive Bayes", fontsize=14)
plt.tight_layout()
plt.savefig("nb_confusion_matrix.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# =========================
# SVM - TUNING VỚI DEV SET
# =========================

svm_params = {"C": [0.1, 1, 10]}

best_score = 0
best_C = None
best_model = None
scores = []
C_used = []

for C in svm_params["C"]:
    # Khởi tạo mô hình với từng giá trị C
    model = LinearSVC(C=C, class_weight='balanced', random_state=42, max_iter=1000)
    
    # Huấn luyện trên Train
    model.fit(X_train_tfidf, y_train)
    
    # Dự đoán trên Dev
    y_dev_pred = model.predict(X_dev_tfidf)
    
    # Tính Macro F1 trên Dev
    score = f1_score(y_dev, y_dev_pred, average='macro')
    
    scores.append(score)
    C_used.append(C)
    
    print(f"C = {C}, Macro F1 on Dev = {score:.4f}")
    
    # Lưu lại mô hình tốt nhất
    if score > best_score:
        best_score = score
        best_C = C
        best_model = model

print(f"\nBest C = {best_C}, Macro F1 on Dev = {best_score:.4f}")
```

```python
# =========================
# VISUALIZE SVM
# =========================

plt.figure(figsize=(8, 5))
plt.plot(C_used, scores, marker='o', linestyle='-', linewidth=2, markersize=8)

plt.xlabel("C", fontsize=12)
plt.ylabel("Macro F1 (Dev)", fontsize=12)
plt.title("SVM - Hyperparameter Tuning", fontsize=14)
plt.xscale('log')  # Vì C thường được khảo sát trên thang log

# Thêm giá trị lên các điểm
for i in range(len(C_used)):
    plt.text(C_used[i], scores[i] + 0.002, f"{scores[i]:.3f}", 
             ha='center', va='bottom', fontsize=10)

plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig("svm_tuning.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
final_svm = LinearSVC(C=best_C, class_weight='balanced', random_state=42, max_iter=1000)
final_svm.fit(X_train_tfidf, y_train)
```

```python
# =========================
# ĐÁNH GIÁ TRÊN TEST
# =========================

y_test_pred = final_svm.predict(X_test_tfidf)

test_accuracy = accuracy_score(y_test, y_test_pred)
test_f1_macro = f1_score(y_test, y_test_pred, average='macro')

print("\n=== KẾT QUẢ TRÊN TEST ===")
print(f"Accuracy = {test_accuracy:.4f}")
print(f"Macro F1 = {test_f1_macro:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_test_pred, digits=4))
```

```python
# =========================
# CONFUSION MATRIX VỚI SEABORN
# =========================

cm = confusion_matrix(y_test, y_test_pred)
labels = ['Tiêu cực', 'Trung tính', 'Tích cực']


plt.figure(figsize=(7, 6))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=labels,
    yticklabels=labels,
    cbar_kws={'label': 'Số lượng mẫu'}
)

plt.xlabel("Predicted Label", fontsize=12)
plt.ylabel("True Label", fontsize=12)
plt.title("Confusion Matrix - SVM", fontsize=14)
plt.tight_layout()
plt.savefig("svm_confusion_matrix.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# =========================
# RANDOM FOREST - TUNING VỚI DEV SET
# =========================

rf_params = {
    "n_estimators": [20, 50, 100, 200],
    "max_depth": [None, 10, 20]
}

best_score = 0
best_params = None
best_model = None
results = []  # Lưu kết quả để vẽ biểu đồ

for n_est in rf_params["n_estimators"]:
    for max_depth in rf_params["max_depth"]:
        # Khởi tạo mô hình với từng bộ tham số
        model = RandomForestClassifier(
            n_estimators=n_est, 
            max_depth=max_depth,
            class_weight='balanced',
            random_state=42,
            n_jobs=-1
        )
        
        # Huấn luyện trên Train
        model.fit(X_train_tfidf, y_train)
        
        # Dự đoán trên Dev
        y_dev_pred = model.predict(X_dev_tfidf)
        
        # Tính Macro F1 trên Dev
        score = f1_score(y_dev, y_dev_pred, average='macro')
        
        results.append({
            'n_estimators': n_est,
            'max_depth': max_depth,
            'macro_f1': score
        })
        
        print(f"n_estimators = {n_est}, max_depth = {max_depth}, Macro F1 on Dev = {score:.4f}")
        
        # Lưu lại mô hình tốt nhất
        if score > best_score:
            best_score = score
            best_params = {'n_estimators': n_est, 'max_depth': max_depth}
            best_model = model

print(f"\nBest params: n_estimators = {best_params['n_estimators']}, max_depth = {best_params['max_depth']}")
print(f"Best Macro F1 on Dev = {best_score:.4f}")
```

```python
# =========================
# VISUALIZE RANDOM FOREST
# =========================

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Lấy giá trị unique
n_est_values = sorted(set([r['n_estimators'] for r in results]))

depth_values = sorted(
    set(r['max_depth'] for r in results),
    key=lambda x: (x is None, x)
)

# Chuyển None -> "None" để hiển thị
depth_values_display = ["None" if d is None else d for d in depth_values]

# Map kết quả để lookup nhanh
result_map = {
    (r['n_estimators'], "None" if r['max_depth'] is None else r['max_depth']): r['macro_f1']
    for r in results
}

# Tạo matrix (dùng NaN thay vì 0)
score_matrix = np.full((len(n_est_values), len(depth_values)), np.nan)

for i, n_est in enumerate(n_est_values):
    for j, depth in enumerate(depth_values):
        depth_key = "None" if depth is None else depth
        score_matrix[i, j] = result_map.get((n_est, depth_key), np.nan)

# Vẽ heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(
    score_matrix,
    annot=True,
    fmt=".4f",
    xticklabels=depth_values_display,
    yticklabels=n_est_values,
    cmap="YlOrRd",
    linewidths=0.5,
    linecolor="gray",
    cbar_kws={'label': 'Macro F1 (Dev)'}
)

plt.xlabel("max_depth", fontsize=12)
plt.ylabel("n_estimators", fontsize=12)
plt.title("Random Forest - Hyperparameter Tuning", fontsize=14)

plt.tight_layout()
plt.savefig("rf_tuning.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
# =========================
# HUẤN LUYỆN LẠI VỚI BEST PARAMS
# =========================

final_rf = RandomForestClassifier(
    n_estimators=best_params['n_estimators'],
    max_depth=best_params['max_depth'],
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)
final_rf.fit(X_train_tfidf, y_train)
```

```python
# =========================
# ĐÁNH GIÁ TRÊN TEST
# =========================

y_test_pred = final_rf.predict(X_test_tfidf)

test_accuracy = accuracy_score(y_test, y_test_pred)
test_f1_macro = f1_score(y_test, y_test_pred, average='macro')

print("\n=== KẾT QUẢ TRÊN TEST ===")
print(f"Accuracy = {test_accuracy:.4f}")
print(f"Macro F1 = {test_f1_macro:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_test_pred, digits=4))
```

```python
# =========================
# CONFUSION MATRIX VỚI SEABORN
# =========================

cm = confusion_matrix(y_test, y_test_pred)
labels = ['Tiêu cực', 'Trung tính', 'Tích cực']


plt.figure(figsize=(7, 6))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=labels,
    yticklabels=labels,
    cbar_kws={'label': 'Số lượng mẫu'}
)

plt.xlabel("Predicted Label", fontsize=12)
plt.ylabel("True Label", fontsize=12)
plt.title("Confusion Matrix - Random Forest", fontsize=14)
plt.tight_layout()
plt.savefig("rf_confusion_matrix.png", dpi=300, bbox_inches='tight')
plt.show()
```
